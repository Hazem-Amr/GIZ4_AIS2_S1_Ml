### Stopwords: list vs set

- `stopwords.words('arabic')` → returns a **list**
- `set(stopwords.words('arabic'))` → converts it to a **set**

#### Difference
- **List:** slower lookup → `word in stop_words` is O(n)
- **Set:** faster lookup → O(1)

#### Conclusion
Use a **set** for filtering stopwords:
```python
stop_words = set(stopwords.words('arabic'))

## stop words in english

````python
import nltk
# nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
````

````python
# Define a sample English text
english_text = "Hello, how are you doing? i hope you are doing well."

# Tokenize the text into words
words = word_tokenize(english_text)
print(words)
print(len(words))
````

````python
# Load the English stop words
stop_words = set(stopwords.words('english'))

# Filter out the stop words from the text
filtered_words = [word for word in words if not word in stop_words]

print(filtered_words)
print(len(filtered_words))
````

````python
# Join the filtered words into a string
filtered_text = ' '.join(filtered_words)

# Print the filtered text
print(filtered_text)

#you can exlude some words in stop of words if you see the result nit good
````

````python
import string

english_text_1 = "Hello, how are you doing? i hope you are doing well."

words_1 = word_tokenize(english_text_1)

# 1. Get the stop words
stop_words_1 = set(stopwords.words('english'))

# 2. Get the punctuation marks
punctuation = set(string.punctuation)

# 3. Combine them or check for both
filtered_words_1 = [word for word in words_1 if word.lower() not in stop_words_1 and word not in punctuation]

filtered_text_1 = " ".join(filtered_words_1)
filtered_text_1
````

# Stemming

````python
import nltk
from nltk.stem import PorterStemmer
````

````python
# Create a stemmer object
stemmer = PorterStemmer()

# Stem some example words
words = ["walking", "jumps", "jumped", "jumping"]
for word in words:
    stemmed_word = stemmer.stem(word)
    print(f"{word} -> {stemmed_word}")
````

#Lemitaziation

````python
import nltk
# nltk.download('wordnet')
````

````python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

text = "The quick brown foxes jumped over the lazy dogs"

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in word_tokenize(text)]
print(stemmed_words)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in word_tokenize(text)]
print(lemmatized_words)

#look for the lazy
````

## search pattern

1- search  >>>>>return the first pattern achieve the condition.

2- findall  >>>>>return all the patterns achieve the condition

3- use groups() to print it

````python
# [('2021 Q1', '4.85'), ('2020 Q4', '3')]

text = '''
Tesla's gross cost of operating lease vehicles in FY2021 Q1 was $4.85 billion.
In previous quarter i.e. FY2020 Q4 it was $3 billion.
'''
pattern = r'FY(\d{4}.Q[1-4])[^\$]+\$([0-9\.]+)'

matches = re.search(pattern, text)
matches.groups()
````
